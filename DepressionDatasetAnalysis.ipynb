```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Load and preprocess the dataset
da = pd.read_csv(r'C:\Users\acpgi\Downloads\BVH\BVH\Combined_csv_file.csv')
da.columns = da.columns.str.split(',').str[0].str.strip()

# Define predictors and group by 'Class'
predictor_columns = [col for col in da.columns if col != 'Class']
classwise_means = da.groupby('Class')[predictor_columns].mean().reset_index()

# Iterate over each class and plot
for _, row in classwise_means.iterrows():
    class_value = row['Class']
    predictors = predictor_columns
    means = row[predictors]
    
    # Create a bar plot for this class
    plt.figure(figsize=(10, 6))
    sns.barplot(x=predictors, y=means)
    
    # Add mean values as labels on each bar
    for i, value in enumerate(means):
        plt.text(i, value + 0.01, f"{value:.2f}", ha='center', va='bottom', fontsize=9)
    
    # Add titles and labels
    plt.title(f'Predictor Means for Class {class_value}')
    plt.xlabel('Predictors')
    plt.ylabel('Mean Value')
    plt.xticks(rotation=45, ha='right')  # Rotate predictor names for better readability
    plt.tight_layout()
    plt.show()
```

```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split, cross_val_score
from sklearn.preprocessing import StandardScaler, LabelEncoder
from sklearn.ensemble import RandomForestClassifier
from sklearn.svm import SVC
from sklearn.metrics import classification_report, accuracy_score

# Step 1: Load and preprocess the dataset
da = pd.read_csv(r'C:\Users\acpgi\Downloads\BVH\BVH\Combined_csv_file.csv')
da.columns = da.columns.str.split(',').str[0].str.strip()

# Check for missing values
if da.isnull().sum().any():
    print("Handling missing values...")
    da.fillna(da.mean(), inplace=True)

# Split features and target
X = da.drop(columns=['Class'])
y = da['Class']

# Encode target variable if necessary
if y.dtype == 'object':
    le = LabelEncoder()
    y = le.fit_transform(y)

# Standardize the predictors
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# Split into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(X_scaled, y, test_size=0.2, random_state=42)

# Step 2: Train multiple models
# Example 1: Random Forest
rf_model = RandomForestClassifier(random_state=42)
rf_model.fit(X_train, y_train)
rf_predictions = rf_model.predict(X_test)

# Example 2: Support Vector Machine
svm_model = SVC(probability=True, random_state=42)
svm_model.fit(X_train, y_train)
svm_predictions = svm_model.predict(X_test)

# Step 3: Evaluate models
print("Random Forest Classification Report:")
print(classification_report(y_test, rf_predictions))
print("Accuracy:", accuracy_score(y_test, rf_predictions))

print("\nSVM Classification Report:")
print(classification_report(y_test, svm_predictions))
print("Accuracy:", accuracy_score(y_test, svm_predictions))

# Optional: Perform cross-validation on Random Forest
cv_scores = cross_val_score(rf_model, X_scaled, y, cv=5)
print("\nRandom Forest Cross-Validation Accuracy:", np.mean(cv_scores))
```

```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

# Load the dataset
da = pd.read_csv(r'C:\Users\acpgi\Downloads\BVH\BVH\Combined_csv_file.csv')
da.columns = da.columns.str.split(',').str[0].str.strip()

# Step 1: Calculate the correlation matrix
correlation_matrix = da.corr()

# Step 2: Plot the correlation matrix to visualize it
plt.figure(figsize=(12, 8))
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', fmt='.2f', linewidths=0.5)
plt.title("Correlation Matrix of Predictors")
plt.show()

# Step 3: Check correlation with the target variable 'Class'
target_corr = correlation_matrix['Class'].sort_values(ascending=False)
print("Correlation with target variable 'Class':")
print(target_corr)

# Step 4: Drop features with low correlation with 'Class'
# Set a threshold value for correlation (e.g., drop features with correlation < 0.1)
threshold = 0.1
low_corr_features = target_corr[target_corr.abs() < threshold].index.tolist()

# Remove the 'Class' column from the list of features to drop
low_corr_features = [feature for feature in low_corr_features if feature != 'Class']

print(f"Features with low correlation (below {threshold}): {low_corr_features}")

# Drop these features
da_reduced = da.drop(columns=low_corr_features)

# Step 5: Check the shape of the reduced dataset
print(f"Reduced dataset shape: {da_reduced.shape}")
```

```python
# Step 6: Remove highly correlated predictors (optional)
threshold = 0.9  # Set the correlation threshold
corr_matrix = da.corr().abs()

# Select upper triangle of the correlation matrix
upper = corr_matrix.where(np.triu(np.ones(corr_matrix.shape), k=1).astype(bool))

# Find index of columns with correlation greater than the threshold
to_drop = [column for column in upper.columns if any(upper[column] > threshold)]

# Drop highly correlated columns
da_reduced = da.drop(columns=to_drop)

print(f"Columns with high correlation (> {threshold}): {to_drop}")
print(f"Reduced dataset shape after removing highly correlated features: {da_reduced.shape}")
```

```python
import pandas as pd
import numpy as np
import lightgbm as lgb
import tensorflow as tf
from tensorflow import keras
from sklearn.model_selection import train_test_split, cross_val_score
from sklearn.preprocessing import StandardScaler, LabelEncoder
from sklearn.ensemble import RandomForestClassifier
from sklearn.svm import SVC
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import classification_report, accuracy_score

# Step 1: Load and preprocess the dataset
da = pd.read_csv(r'C:\Users\acpgi\Downloads\BVH\BVH\Combined_csv_file.csv')
da.columns = da.columns.str.split(',').str[0].str.strip()

# Check for missing values and handle them
if da.isnull().sum().any():
    print("Handling missing values...")
    da.fillna(da.mean(), inplace=True)

# Split features and target
X = da.drop(columns=['Class'])
y = da['Class']

# Encode target variable if necessary
if y.dtype == 'object':
    le = LabelEncoder()
    y = le.fit_transform(y)

# Standardize the predictors
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# Split into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(X_scaled, y, test_size=0.2, random_state=42)

# Step 2: Train multiple models

# Logistic Regression
log_reg = LogisticRegression(random_state=42, max_iter=1000)
log_reg.fit(X_train, y_train)
log_reg_predictions = log_reg.predict(X_test)

# Decision Tree Classifier
dt_model = DecisionTreeClassifier(random_state=42)
dt_model.fit(X_train, y_train)
dt_predictions = dt_model.predict(X_test)

# k-Nearest Neighbors (KNN)
knn_model = KNeighborsClassifier(n_neighbors=5)
knn_model.fit(X_train, y_train)
knn_predictions = knn_model.predict(X_test)

# Gradient Boosting (LightGBM)
lgb_model = lgb.LGBMClassifier(random_state=42)
lgb_model.fit(X_train, y_train)
lgb_predictions = lgb_model.predict(X_test)

# Artificial Neural Network (ANN) using TensorFlow/Keras
ann_model = keras.Sequential([
    keras.layers.Dense(64, activation='relu', input_shape=(X_train.shape[1],)),
    keras.layers.Dense(32, activation='relu'),
    keras.layers.Dense(len(np.unique(y)), activation='softmax')  # Softmax for multi-class classification
])

ann_model.compile(optimizer='adam', loss='sparse_categorical_crossentropy', metrics=['accuracy'])
ann_model.fit(X_train, y_train, epochs=30, batch_size=32, verbose=0)
ann_predictions = np.argmax(ann_model.predict(X_test), axis=1)

# Random Forest
rf_model = RandomForestClassifier(random_state=42)
rf_model.fit(X_train, y_train)
rf_predictions = rf_model.predict(X_test)

# Support Vector Machine (SVM)
svm_model = SVC(probability=True, random_state=42)
svm_model.fit(X_train, y_train)
svm_predictions = svm_model.predict(X_test)

# Step 3: Evaluate models
models = {
    "Logistic Regression": log_reg_predictions,
    "Decision Tree": dt_predictions,
    "k-NN": knn_predictions,
    "LightGBM": lgb_predictions,
    "ANN": ann_predictions,
    "Random Forest": rf_predictions,
    "SVM": svm_predictions
}

for model_name, predictions in models.items():
    print(f"\n{model_name} Classification Report:")
    print(classification_report(y_test, predictions))
    print(f"Accuracy: {accuracy_score(y_test, predictions):.4f}")

# Cross-validation on Random Forest as an example
cv_scores = cross_val_score(rf_model, X_scaled, y, cv=5)
print(f"\nRandom Forest Cross-Validation Accuracy: {np.mean(cv_scores):.4f}")
```

```python
import pandas as pd
import numpy as np

# Load dataset
da = pd.read_csv(r'C:\Users\acpgi\Downloads\BVH\BVH\Combined_csv_file.csv')
da.columns = da.columns.str.split(',').str[0].str.strip()

# Encode target variable if necessary
if da['Class'].dtype == 'object':
    from sklearn.preprocessing import LabelEncoder
    le = LabelEncoder()
    da['Class'] = le.fit_transform(da['Class'])

# Compute correlation matrix
correlation_matrix = da.corr()

# Correlation of features with target variable
target_corr = correlation_matrix['Class'].drop('Class')

# Define threshold
threshold = 0.2

# Select features with correlation above threshold
selected_features = target_corr[(target_corr.abs() >= threshold)].index.tolist()

# Drop low-correlation features
X_filtered = da[selected_features]
y = da['Class']

print(f"Selected {len(selected_features)} features based on correlation threshold of ±{threshold}.")
print("Remaining Features:", selected_features)

# Save filtered dataset
filtered_df = pd.concat([X_filtered, y], axis=1)
filtered_df.to_csv('filtered_dataset.csv', index=False)
```

```python
import pandas as pd
import numpy as np
import lightgbm as lgb
import tensorflow as tf
from tensorflow import keras
from sklearn.model_selection import train_test_split, cross_val_score
from sklearn.preprocessing import StandardScaler, LabelEncoder
from sklearn.ensemble import RandomForestClassifier
from sklearn.svm import SVC
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import classification_report, accuracy_score

# Step 1: Load and preprocess the dataset
da = pd.read_csv(r'C:\Users\acpgi\filtered_dataset.csv')
da.columns = da.columns.str.split(',').str[0].str.strip()

# Check for missing values and handle them
if da.isnull().sum().any():
    print("Handling missing values...")
    da.fillna(da.mean(), inplace=True)

# Split features and target
X = da.drop(columns=['Class'])
y = da['Class']

# Encode target variable if necessary
if y.dtype == 'object':
    le = LabelEncoder()
    y = le.fit_transform(y)

# Standardize the predictors
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# Split into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(X_scaled, y, test_size=0.2, random_state=42)

# Step 2: Train multiple models

# Logistic Regression
log_reg = LogisticRegression(random_state=42, max_iter=1000)
log_reg.fit(X_train, y_train)
log_reg_predictions = log_reg.predict(X_test)

# Decision Tree Classifier
dt_model = DecisionTreeClassifier(random_state=42)
dt_model.fit(X_train, y_train)
dt_predictions = dt_model.predict(X_test)

# k-Nearest Neighbors (KNN)
knn_model = KNeighborsClassifier(n_neighbors=5)
knn_model.fit(X_train, y_train)
knn_predictions = knn_model.predict(X_test)

# Gradient Boosting (LightGBM)
lgb_model = lgb.LGBMClassifier(random_state=42)
lgb_model.fit(X_train, y_train)
lgb_predictions = lgb_model.predict(X_test)

# Artificial Neural Network (ANN) using TensorFlow/Keras
ann_model = keras.Sequential([
    keras.layers.Dense(64, activation='relu', input_shape=(X_train.shape[1],)),
    keras.layers.Dense(32, activation='relu'),
    keras.layers.Dense(len(np.unique(y)), activation='softmax')  # Softmax for multi-class classification
])

ann_model.compile(optimizer='adam', loss='sparse_categorical_crossentropy', metrics=['accuracy'])
ann_model.fit(X_train, y_train, epochs=30, batch_size=32, verbose=0)
ann_predictions = np.argmax(ann_model.predict(X_test), axis=1)

# Random Forest
rf_model = RandomForestClassifier(random_state=42)
rf_model.fit(X_train, y_train)
rf_predictions = rf_model.predict(X_test)

# Support Vector Machine (SVM)
svm_model = SVC(probability=True, random_state=42)
svm_model.fit(X_train, y_train)
svm_predictions = svm_model.predict(X_test)

# Step 3: Evaluate models
models = {
    "Logistic Regression": log_reg_predictions,
    "Decision Tree": dt_predictions,
    "k-NN": knn_predictions,
    "LightGBM": lgb_predictions,
    "ANN": ann_predictions,
    "Random Forest": rf_predictions,
    "SVM": svm_predictions
}

for model_name, predictions in models.items():
    print(f"\n{model_name} Classification Report:")
    print(classification_report(y_test, predictions))
    print(f"Accuracy: {accuracy_score(y_test, predictions):.4f}")

# Cross-validation on Random Forest as an example
cv_scores = cross_val_score(rf_model, X_scaled, y, cv=5)
print(f"\nRandom Forest Cross-Validation Accuracy: {np.mean(cv_scores):.4f}")
```
